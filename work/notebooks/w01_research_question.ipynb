## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
## 2. The question
print("""Research question: Which content pages should FlyRank's team review first for a refresh,
given limited reviewer time?

Unit of analysis: one page (content_id), scored and ranked at a point in time.

Output: a ranked queue of pages, each with a score and a reason code explaining why it
was flagged (e.g. "stale + visible", "declining with demand").

The action: a content reviewer opens the top of the queue and manually reviews/updates
those pages first, instead of reviewing pages in an arbitrary or gut-feel order.

Cost of a wrong call:
- False positive (flagged high, but not actually worth reviewing): wastes a reviewer's time —
  the opportunity cost is whatever else they could have reviewed instead.
- False negative (a genuinely declining/high-value page ranked low): the page keeps losing
  visibility/traffic without anyone noticing, potentially for months.
- Because reviewer time is the scarce resource, precision at the top of the list (Precision@K)
  matters more than perfect accuracy across all pages — a bad #3 ranking costs more than a bad
  #3000 ranking.

Why data/ML can help: with thousands of pages and limited reviewer capacity, someone has
to decide an order. A hand-written rule (like "stale + visible") is transparent but blunt — the
starter pipeline showed a learned model roughly tripling the hand rule's Precision@50 on this
dataset. That gap suggests there's real signal in combinations of features that a human
wouldn't naturally weight by hand.""")
```
